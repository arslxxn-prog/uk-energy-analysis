```python
from pathlib import Path
import pandas as pd

file_path = Path("../data/raw/fuelhh/fuelhh_2025.csv")

df = pd.read_csv(file_path)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print("\nColumn names:")
print(df.columns.tolist())
```

```python
df["settlementDate"] = pd.to_datetime(df["settlementDate"])

print("Minimum date:", df["settlementDate"].min())
print("Maximum date:", df["settlementDate"].max())
print("Number of unique dates:", df["settlementDate"].nunique())
```

```python
print("Settlement periods:")
print(sorted(df["settlementPeriod"].unique()))

print("\nNumber of unique settlement periods:")
print(df["settlementPeriod"].nunique())
```

```python
fuel_types = sorted(df["fuelType"].unique())

print(f"Number of fuel categories: {len(fuel_types)}")

for fuel in fuel_types:
    print(f"- {fuel}")
```

```python
print(df.isna().sum())
```

```python
key_columns = [
    "settlementDate",
    "settlementPeriod",
    "fuelType"
]

duplicate_count = df.duplicated(
    subset=key_columns
).sum()

print(f"Duplicate records: {duplicate_count:,}")
```

```python
expected_rows = (
    df["settlementDate"].nunique()
    * df["settlementPeriod"].nunique()
    * df["fuelType"].nunique()
)

actual_rows = len(df)

print(f"Expected rows: {expected_rows:,}")
print(f"Actual rows:   {actual_rows:,}")
print(f"Difference:    {actual_rows - expected_rows:,}")
```

```python
print(df["generation"].describe())

negative_count = (df["generation"] < 0).sum()

print(f"\nNegative generation values: {negative_count:,}")
```
