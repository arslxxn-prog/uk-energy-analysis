```python
from pathlib import Path
import pandas as pd

data_dir = Path("../data/raw/fuelhh")
files = sorted(data_dir.glob("fuelhh_*.csv"))

key_columns = [
    "settlementDate",
    "settlementPeriod",
    "fuelType"
]

for file in files:
    df_year = pd.read_csv(file)

    duplicates = (
        df_year[
            df_year.duplicated(
                subset=key_columns,
                keep=False
            )
        ]
        .sort_values(
            key_columns + ["startTime", "publishTime"]
        )
    )

    if not duplicates.empty:
        print("=" * 80)
        print(file.name)
        print(f"Duplicate rows: {len(duplicates):,}")
        print(
            f"Duplicate groups: "
            f"{duplicates.groupby(key_columns).ngroups:,}"
        )
        print()

        print(
            duplicates[
                key_columns
                + ["publishTime", "startTime", "generation"]
            ]
            .head(20)
            .to_string(index=False)
        )
```

```python
for file in files:
    df_year = pd.read_csv(file)

    periods_per_day = (
        df_year
        .groupby("settlementDate")["settlementPeriod"]
        .nunique()
    )

    incomplete = periods_per_day[
        ~periods_per_day.isin([46, 48, 50])
    ]

    if not incomplete.empty:
        print("=" * 80)
        print(file.name)
        print("Dates with unexpected settlement-period counts:")
        print(incomplete.to_string())
```

```python
for file in files:
    df_year = pd.read_csv(file)

    periods_per_day = (
        df_year
        .groupby("settlementDate")["settlementPeriod"]
        .nunique()
    )

    incomplete_dates = periods_per_day[
        ~periods_per_day.isin([46, 48, 50])
    ].index

    if len(incomplete_dates) > 0:
        example_date = incomplete_dates[0]

        print("=" * 80)
        print(f"{file.name} - example date: {example_date}")

        example = df_year[
            df_year["settlementDate"] == example_date
        ]

        print(
            "Settlement periods present:"
        )
        print(
            sorted(
                example["settlementPeriod"].unique()
            )
        )

        print(
            f"Rows returned: {len(example):,}"
        )
    else:
        print(f"{file.name}: no incomplete dates")
```

```python
for file in files:
    df_year = pd.read_csv(file)

    df_year["publishTime"] = pd.to_datetime(
        df_year["publishTime"],
        utc=True
    )

    df_year["startTime"] = pd.to_datetime(
        df_year["startTime"],
        utc=True
    )

    # Sort so the newest publication is first
    df_latest = (
        df_year
        .sort_values("publishTime")
        .drop_duplicates(
            subset=["startTime", "fuelType"],
            keep="last"
        )
    )

    remaining_duplicates = df_latest.duplicated(
        subset=["startTime", "fuelType"]
    ).sum()

    print(
        f"{file.name}: "
        f"{len(df_year):,} raw rows → "
        f"{len(df_latest):,} canonical rows | "
        f"remaining duplicates: "
        f"{remaining_duplicates:,}"
    )
```

```python
for file in files:
    df_year = pd.read_csv(file)

    df_year["publishTime"] = pd.to_datetime(
        df_year["publishTime"],
        utc=True
    )

    df_year["startTime"] = pd.to_datetime(
        df_year["startTime"],
        utc=True
    )

    df_latest = (
        df_year
        .sort_values("publishTime")
        .drop_duplicates(
            subset=["startTime", "fuelType"],
            keep="last"
        )
    )

    periods_per_day = (
        df_latest
        .groupby("settlementDate")["settlementPeriod"]
        .nunique()
    )

    unexpected = periods_per_day[
        ~periods_per_day.isin([46, 48, 50])
    ]

    print("=" * 80)
    print(file.name)

    if unexpected.empty:
        print("No unexpected settlement-period counts.")
    else:
        print(unexpected.to_string())
```

```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path("../data/raw/fuelhh")
PROCESSED_DIR = Path("../data/processed")

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

files = sorted(RAW_DIR.glob("fuelhh_*.csv"))

dfs = []

for file in files:
    print(f"Loading {file.name}...")
    
    df_year = pd.read_csv(
        file,
        dtype={
            "dataset": "string",
            "fuelType": "string"
        }
    )
    
    dfs.append(df_year)

fuelhh = pd.concat(
    dfs,
    ignore_index=True
)

print(f"\nTotal rows: {len(fuelhh):,}")
print(f"Columns: {fuelhh.columns.tolist()}")
```

```python
fuelhh["startTime"] = pd.to_datetime(
    fuelhh["startTime"],
    utc=True
)

fuelhh["publishTime"] = pd.to_datetime(
    fuelhh["publishTime"],
    utc=True
)
```

```python
fuelhh["datetime_uk"] = (
    fuelhh["startTime"]
    .dt.tz_convert("Europe/London")
)
```

```python
duplicate_key_count = fuelhh.duplicated(
    subset=["startTime", "fuelType"]
).sum()

print(
    f"Duplicate startTime + fuelType rows: "
    f"{duplicate_key_count:,}"
)
```

```python
fuel_profile = (
    fuelhh
    .groupby("fuelType")
    .agg(
        records=("fuelType", "size"),
        min_generation_mw=("generation", "min"),
        max_generation_mw=("generation", "max"),
        mean_generation_mw=("generation", "mean"),
    )
    .sort_index()
)

fuel_profile
```

```python
fuelhh["date"] = fuelhh["datetime_uk"].dt.date
fuelhh["hour"] = fuelhh["datetime_uk"].dt.hour
fuelhh["minute"] = fuelhh["datetime_uk"].dt.minute
fuelhh["day_of_week"] = fuelhh["datetime_uk"].dt.day_name()
fuelhh["month"] = fuelhh["datetime_uk"].dt.month
fuelhh["year"] = fuelhh["datetime_uk"].dt.year
fuelhh["is_weekend"] = (
    fuelhh["datetime_uk"].dt.dayofweek >= 5
)
```
