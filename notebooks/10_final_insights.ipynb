```python
from pathlib import Path
import pandas as pd
import numpy as np

file_path = Path("../data/processed/gb_energy_analysis.csv")

df = pd.read_csv(file_path)

print(f"Rows: {len(df):,}")
print(f"Columns: {len(df.columns)}")
print(df.columns.tolist())
```

```python
year_summary = (
    df.groupby("year")
      .agg(
          average_demand_mw=("demand_mw", "mean"),
          peak_demand_mw=("demand_mw", "max"),
          average_gas_mw=("gas_mw", "mean"),
          average_renewable_mw=("renewable_generation_mw", "mean"),
          average_wind_mw=("wind_mw", "mean"),
          average_net_demand_mw=("net_demand_mw", "mean")
      )
      .round(2)
)

year_summary
```

```python
year_summary = (
    df.groupby("year")
      .agg(
          average_demand_mw=("demand_mw", "mean"),
          peak_demand_mw=("demand_mw", "max"),
          average_gas_mw=("gas_mw", "mean"),
          average_renewable_mw=("renewable_generation_mw", "mean"),
          average_wind_mw=("wind_mw", "mean"),
          average_net_demand_mw=("net_demand_mw", "mean")
      )
      .round(2)
)

year_summary
```

```python
renewable_share = (
    df.groupby("year")
      .agg(
          renewable_mw=("renewable_generation_mw", "mean"),
          demand_mw=("demand_mw", "mean")
      )
)

renewable_share["renewable_share"] = (
    renewable_share["renewable_mw"] /
    renewable_share["demand_mw"]
)

renewable_share["renewable_share_pct"] = (
    renewable_share["renewable_share"] * 100
)

renewable_share.round(2)
```

```python
peak_year = year_summary["peak_demand_mw"].idxmax()
peak_value = year_summary["peak_demand_mw"].max()

print(f"Peak demand year: {peak_year}")
print(f"Peak demand: {peak_value:,.0f} MW")
```

```python
hourly_demand = (
    df.groupby("hour")["demand_mw"]
      .mean()
      .sort_values(ascending=False)
)

print("Highest average-demand hours:")
print(hourly_demand.head(5).round(0))

print("\nLowest average-demand hours:")
print(hourly_demand.sort_values().head(5).round(0))
```

```python
net_hourly = (
    df.groupby("hour")["net_demand_mw"]
      .mean()
      .sort_values(ascending=False)
)

print("Highest average net-demand hours:")
print(net_hourly.head(5).round(0))
```
